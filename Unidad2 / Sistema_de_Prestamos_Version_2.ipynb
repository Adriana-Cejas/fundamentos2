```python
prestamos = []
import datetime
import pytz
necesidades = []
# ========================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ========================================

# Lista de alumnos activos en el sistema

alumnos = [
          [ "26040200", "Luis", "Mecanica"],
          [ "26040201", "Alonso", "Mecatronica"],
          ["26040202", "Adriana", "Inteligencia Artificial"],
          ["26040203", "Mariana", "Electronica"],
          ["26040204", "Lakas", "Arquitectura"],
         ]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras =[
               ["PC-LC1-01", "LC1" ,"Python, VS Code", "Disponible"],
               ["PC-LC1-02", "LC1" ,"Python, Java, VS Code", "Disponible"],
               ["PC-LC1-03", "LC1" ,"AutoCAD, MATLAB", "Reservada"],
               ["PC-LC2-01", "LC2" ,"Python, VS Code", "Disponible"],
               ["PC-LC2-02", "LC2" ,"Python, Java, VS Code", "Prestada"],
               ["PC-LC2-03", "LC2" , "AutoCAD, MATLAB", "Reservada"],
               ["PC-LC3-01", "LC3" ,"Python, VS Code", "Disponible"],
               ["PC-LC3-02", "LC3" ,"Python, Java, VS Code", "Prestada"],
               ["PC-LC3-03", "LC3" , "AutoCAD, MATLAB", "Reservada"],
               ["PC-LC3-04", "LC3" ,"Python, VS Code", "Mantenimiento" ],
              ]

def consultar_computadoras():
  print("\n--- COMPUTADORAS ---")
  for i in range(len(computadoras)):
    print("ID:",computadoras[i][0],"| Aula:", computadoras[i][1] ,
          "| Software:", computadoras[i][2]," |Estado:", computadoras[i][3] )

def buscar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
        print("ID:", computadoras[i][0], "| Aula:", computadoras[i][1],
              "| Software:", computadoras[i][2],"| Estado:", computadoras[i][3])
        encontrado = True

  if not encontrado:
    print("No se encontraron computadoras con ese software.")

def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("Dame el ID de la computadora:")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
          computadoras[i][3] = "Disponible"
          print(f"La computadora, {id_pc}, ya está disponible.")
          break
      else:
        print(f"La computadora, {id_pc}, ya está disponible.")
  if not existe:
    print(f"No existe la computadora con ID, {id_pc}.")

def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  computadoras.append([id_computadora, aula, programas, "Disponible"])
  print("\nComputadora agregada correctamente.")

def solicitar_prestamo():
   print("\n===SOLICITAR PRÉSTAMO===")
   num_control= input("Número de control del alumno:")
   alumno_encontrado = False
   for i in range(len(alumnos)):
    if alumnos[i][0] == num_control:
      alumno_encontrado = True
      print(f"Alumno encontrado: {alumnos[i][1]} ({alumnos[i][2]})")
      break
   if  not alumno_encontrado:
    print("No se encontro a ese alumno.")
    return

   programa = input("¿Que programa necesita?:")
   computadora_asignada = False
   for i in range(len(computadoras)):
     current_id = computadoras[i][0]
     current_aula = computadoras[i][1]
     current_software = computadoras[i][2]
     current_estado = computadoras[i][3]

     if programa.lower() in current_software.lower():
       if current_estado == "Disponible":
         computadoras [i][3]="Prestada"
         print(f"Se asignó la computadora {current_id} (Aula: {current_aula}) que tiene'{current_software}'.")
         prestamo = [num_control,programa,current_aula,current_id]
         prestamos.append(prestamo)
         computadora_asignada = True
         break
       else:
         print(f"La computadora {current_id} tiene el software '{current_software}' pero está '{current_estado}'.")

   if not computadora_asignada:
    print(f"No se encontró una computadora disponible con el software '{programa}'.")
    confirmar = input("¿Desea agregar este programa a la lista de necesidades? (s/n): ")
    if confirmar.lower() == 's':
      tz_mexico = pytz.timezone('America/Mexico_City')
      fecha_actual = datetime.datetime.now(tz_mexico).strftime("%d-%m-%Y %H:%M:%S")
      necesidades.append([programa, fecha_actual])
      print("Programa agregado a la lista de necesidades exitosamente.")

def historial_prestamos():
  print("\n--- HISTORIAL DE  PRESTAMOS DE COMPUTADORAS ---")
  if not prestamos:
    print("No hay préstamos registrados en el historial.")
    return
  for i in range(len(prestamos)):
    print("numero de control:",prestamos[i][0],"| Programa:", prestamos[i][1] ,
          "| Aula:", prestamos[i][2]," | ID:", prestamos[i][3] )

def historial_necesidades():
  print("\n--- HISTORIAL DE NACESIDADES ---")
  if not necesidades:
    print("No hay necesidades de software registradas.")
  else:
    for i in range(len(necesidades)):
      print(f"{i + 1} | Programa: {necesidades[i][0]} | Fecha: {necesidades[i][1]} | ")

# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 8 (salir) o cualquier numero superior a 7
opcion = 0
while opcion !=10:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de Préstamos")
    print("8. Necesidades de Software")
    print("9. Estadísticas por carrera y Programa")
    print("10. SALIR")
    print("================================")
    try:
        opcion = int(input("Seleccione una opción: "))
    except ValueError:
        print("\nEntrada inválida. Por favor, ingrese un número.")
        continue

    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
      consultar_computadoras()

    # 2. Buscar por software
    elif opcion == 2:
      buscar_sw()

    elif opcion ==3:
     solicitar_prestamo()

    elif opcion==4:
      devolver_computadora()

    # 6. Agregar computadora
    elif opcion == 6:
      agregar_computadora()

    elif opcion ==7:
     historial_prestamos()

    elif opcion == 8:
      historial_necesidades()

    elif opcion == 9:
      print("\n---ESTADÍSTICA POR CARRERA ---")
      conteo_carreras = []
      for i in range(len(prestamos)):
        num_control = prestamos[i][0]
        carrera = ""
        #Buscar la carrera en la lista de alumnos
        for j in range(len(alumnos)):
          if alumnos[j][0] == num_control:
            carrera = alumnos[j][2]
        # Contar la carrera
        encontrado = False
        for k in range(len(conteo_carreras)):
          if conteo_carreras[k][0] == carrera:
            conteo_carreras[k][1] += 1
            encontrado = True
            break
        if not encontrado:
          conteo_carreras.append([carrera, 1])

      if not conteo_carreras:
        print("No hay préstamos para mostrar estadísticas.")
      else:
        for k in range(len(conteo_carreras)):
          print(f"Carrera: {conteo_carreras[k][0]} | Cantidad: {conteo_carreras[k][1]}")

      conteo_programas = []
      for i in range(len(prestamos)):
        programa = prestamos[i][1]
        encontrado = False
        #Buscar si ya está en la lista de conteo
        for j in range(len(conteo_programas)):
          if conteo_programas[j][0] == programa:
            conteo_programas[j][1] += 1
            encontrado = True
        # Si no estaba, lo agregamos
        if not encontrado:
          conteo_programas.append([programa, 1])

      print("\nESTADÍSTICAS POR PROGRAMA:")
      for i in range(len(conteo_programas)):
        print(conteo_programas[i][0], ":", conteo_programas[i][1], "veces" )

    # 10. Salir
    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
        break
    else:
        print("\nOpción no válida.")
```
