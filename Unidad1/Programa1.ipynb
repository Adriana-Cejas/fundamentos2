```python
#=========================================
#SISTEMA DE PRESTAMOS
#Version 1 - Programacion basica y uso de listas
#=========================================

#Lista de alumnos activos en el sistema
alumnos = ["2602200", "26020201","26040202","26040203","26040204"]

#Cada valor en la lista de computadora tiene su lista correspondiente con el
#aula donde esta ubicada, el software instalado y el estado de esa computadora
computadora = ["PC-LC1-01","PC-LC1-02","PC-LC1-03",
               "PC-LC2-01","PC-LC2-02","PC-LC2-03",
               "PC-LC3-01","PC-LC3-02","PC-LC3-03","PC-LC3-04"]

aulas = ["LC1","LC1","LC1",
         "LC2","LC2","LC2",
         "LC3","LC3","LC3","LC3"]

software=["Python, VS code","Python,Java,VS code","AutoCAD, MATALAB",
          "Python, VS code","Python,Java,VS code","AutoCAD, MATALAB",
          "Python, VS code","Python,Java,VS code","AutoCAD, MATALAB","Python, VS code"]

estado=["Disponible","Prestada","Reservada",
        "Disponible","Prestada","Reservada",
        "Disponible","Prestada","Reservada","Mantenimiento"]

#MENU PRINCIPAL
#Usamos un ciclo while que se repetira indefinidamente hasta que el usuario
#teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 6:
  print("\n================================")
  print("SISTEMA DE PRESTAMOS")
  print("==================================")
  print("1.Consultar computadoras")
  print("2.Buscar por software")
  print("3.Solicitar prestamo")
  print("4.Devolver computadora")
  print("5.Crear reservacion")
  print("6.Agregar computadora")
  print("7.Salir")
  print("=================================")
  try:
    opcion = int(input("Seleccione una opcion:"))
  except ValueError:
    print("\n¡Error! Por favor, ingrese un número del 1 al 7.")
    continue # Continuar al inicio del bucle para pedir la opción de nuevo

  #1. Consultar computadora
  #si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
  # toda la lista de computadoras e imprime los valores de las demas listas
  if opcion == 1:
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadora)): # Corregido: computadora (singular)
      print("Aula:", aulas[i]) # Corregido: aulas[i]
      print("ID:", computadora[i]) # Corregido: computadora (singular)
      print("Software:", software[i])
      print("Estado:", estado[i]) # Corregido: estado (singular)
      print("------------------------")
  #2.Buscar por software
  elif opcion == 2:
    programa_buscado = input("\nIngrese el software que desea buscar:") # Corregido: variable name
    print("\n--- RESULTADOS ---") # Corregido: Indentación
    encontrado = False # Corregido: Indentación
    for i in range(len(computadora)): # Corregido: Indentación y computadora (singular)
      #la funcion (lower) convierte a minuscula el texto, de esta forma
      #nos aseguramos que el texto a buscar y la informacion de las listas
      #este en minusculas y no tengamos problema de que no encuentra un programa
      #porque este escrito de forma diferente usando mayusculas y minusculas
      if programa_buscado.lower() in software[i].lower(): # Corregido: variable name
        print("ID:",computadora[i]) # Corregido: computadora (singular)
        print("Aula:", aulas[i])
        print("Estado:",estado[i]) # Corregido: estado (singular)
        encontrado = True

    if not encontrado: # Corregido: Simplificado la condición
      print("No se encontraron computadoras con ese software.")
```
